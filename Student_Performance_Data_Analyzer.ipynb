```python
from collections import defaultdict, Counter


LAST_NAME = "Barrientos"
SEED_NUM = 4
FAVORITE_ARTIST = "Mac DeMarco"


students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]

subjects = [
    "PROGRAMMING",
    "MATHEMATICS",
    "ELECTRONICS",
    "PHYSICS",
    "ENGLISH"
]


artist_score = sum(
    (i + 1) * ord(char)
    for i, char in enumerate(FAVORITE_ARTIST)
)

records = []

for i, student in enumerate(students):
    for j, subject in enumerate(subjects):
        score = (SEED_NUM + artist_score + (i + 1) * 7
                 + (j + 1) * 5) % 51 + 50

        records.append({
            "Student": student,
            "Subject": subject,
            "Score": score
        })


records.append(records[0].copy())


records.append({
    "Student": students[0],
    "Subject": "MATHEMATICS",
    "Score": None
})


def get_category(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 80:
        return "GOOD"
    elif score >= 75:
        return "SATISFACTORY"
    else:
        return "NEEDS IMPROVEMENT"

for record in records:
    record["Category"] = get_category(record["Score"])


print("1. GENERATED PERFORMANCE RECORDS")
for record in records:
    print(record)


organized = defaultdict(list)

for record in records:
    organized[record["Student"]].append(record)

print("\n2. ORGANIZED STUDENT/SUBJECT DATA")
for student, data in sorted(organized.items()):
    print(f"\n{student}:")
    for record in data:
        print(record)


seen = set()
repeated = []
incomplete = []

for record in records:
    key = (record["Student"], record["Subject"])

    if key in seen:
        repeated.append(record)
    else:
        seen.add(key)

    if record["Score"] is None:
        incomplete.append(record)

print("\n3. REPEATED/INCOMPLETE RECORDS")
print("Repeated records:", repeated)
print("Incomplete records:", incomplete)


print("\n4. PERFORMANCE CATEGORIES")
category_counts = Counter(
    record["Category"] for record in records
)
for category, count in sorted(category_counts.items()):
    print(category, ":", count)


print("\n5. STUDENT AND SUBJECT SUMMARIES")

print("\nStudent summaries:")
for student in students:
    scores = [
        r["Score"] for r in records
        if r["Student"] == student and r["Score"] is not None
    ]
    print(
        student,
        "| Valid scores:", len(scores),
        "| Average:", round(sum(scores) / len(scores), 2)
    )

print("\nSubject summaries:")
for subject in subjects:
    scores = [
        r["Score"] for r in records
        if r["Subject"] == subject and r["Score"] is not None
    ]
    print(
        subject,
        "| Valid scores:", len(scores),
        "| Average:", round(sum(scores) / len(scores), 2)
    )


print("\n6. ORDERED RESULTS")
ordered = sorted(
    records,
    key=lambda r: (
        r["Student"],
        r["Subject"],
        r["Score"] is None,
        r["Score"] if r["Score"] is not None else 0
    )
)

for record in ordered:
    print(record)


print("\nFINAL STUDENT PERFORMANCE REPORT")
print("Student:", LAST_NAME)
print("Seed number:", SEED_NUM)
print("Favorite artist:", FAVORITE_ARTIST)
print("Total records:", len(records))
print("Repeated records:", len(repeated))
print("Incomplete records:", len(incomplete))

valid_scores = [
    r["Score"] for r in records if r["Score"] is not None
]
print("Overall average:", round(sum(valid_scores) / len(valid_scores), 2))
print("Performance categories:", dict(category_counts))
print("Final student performance report generated successfully.")
```
