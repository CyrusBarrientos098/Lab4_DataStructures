```python
from collections import defaultdict, Counter


LAST_NAME = "Barrientos"
SEED_NUM = 4
FAVORITE_ARTIST = "Mac DeMarco"


customers = [
    f"{LAST_NAME}_C{i}" for i in range(1, 6)
]

products = [
    "NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"
]


prices = {
    "NOTEBOOK": 50,
    "PEN": 15,
    "USB": 350,
    "HEADPHONES": 500,
    "CALCULATOR": 250
}


artist_score = sum(
    (i + 1) * ord(char)
    for i, char in enumerate(FAVORITE_ARTIST)
)

transactions = []

for i, customer in enumerate(customers):
    for j, product in enumerate(products):
        quantity = (
            SEED_NUM + artist_score + (i + 1) * 3 + (j + 1) * 2
        ) % 5 + 1

        transactions.append({
            "Customer": customer,
            "Product": product,
            "Quantity": quantity,
            "Price": prices[product],
            "Total": quantity * prices[product]
        })


transactions.append(transactions[0].copy())


print("1. GENERATED TRANSACTION RECORDS")
for transaction in transactions:
    print(transaction)


organized = defaultdict(list)

for transaction in transactions:
    organized[transaction["Customer"]].append(transaction)

print("\n2. ORGANIZED TRANSACTION DATA")
for customer, data in sorted(organized.items()):
    print(f"\n{customer}:")
    for transaction in data:
        print(transaction)


purchase_counts = Counter(
    (t["Customer"], t["Product"]) for t in transactions
)

repeated = {
    purchase: count
    for purchase, count in purchase_counts.items()
    if count > 1
}

print("\n3. REPEATED/DISTINCT TRANSACTIONS")
print("Total transactions:", len(transactions))
print("Distinct customer-product pairs:", len(purchase_counts))
print("Distinct customers:", len(set(t["Customer"] for t in transactions)))
print("Distinct products:", len(set(t["Product"] for t in transactions)))
print("Repeated purchases:", repeated)


print("\n4. TRANSACTION VALUES")
for transaction in transactions:
    print(
        transaction["Customer"],
        "|", transaction["Product"],
        "| Quantity:", transaction["Quantity"],
        "| Total:", transaction["Total"]
    )

print("Overall transaction value:", sum(t["Total"] for t in transactions))


print("\n5. CUSTOMER SUMMARIES")
for customer in customers:
    data = [t for t in transactions if t["Customer"] == customer]
    print(
        customer,
        "| Transactions:", len(data),
        "| Total quantity:", sum(t["Quantity"] for t in data),
        "| Total spending:", sum(t["Total"] for t in data)
    )


print("\n6. PRODUCT SUMMARIES")
for product in products:
    data = [t for t in transactions if t["Product"] == product]
    print(
        product,
        "| Units sold:", sum(t["Quantity"] for t in data),
        "| Transactions:", len(data),
        "| Total sales:", sum(t["Total"] for t in data)
    )


print("\n7. ORDERED RESULTS")
ordered = sorted(
    transactions,
    key=lambda t: (t["Customer"], t["Product"])
)

for transaction in ordered:
    print(transaction)


print("\n8. OVERALL TRANSACTION SUMMARY")
print("Total transactions:", len(transactions))
print("Total customers:", len(set(t["Customer"] for t in transactions)))
print("Total products:", len(set(t["Product"] for t in transactions)))
print("Total units sold:", sum(t["Quantity"] for t in transactions))
print("Total sales:", sum(t["Total"] for t in transactions))
print("Repeated customer-product purchases:", len(repeated))


print("\nFINAL CONSOLIDATED TRANSACTION REPORT")
print("Store Transaction Analysis")
print("Student:", LAST_NAME)
print("Seed number:", SEED_NUM)
print("Favorite artist:", FAVORITE_ARTIST)
print("Customers analyzed:", ", ".join(customers))
print("Products analyzed:", ", ".join(products))
print("Total transaction records:", len(transactions))
print("Total sales value:", sum(t["Total"] for t in transactions))
print("Distinct customer-product pairs:", len(purchase_counts))
print("Repeated purchases identified:", len(repeated))
print("Transaction report generated successfully.")
```
