```python
from collections import Counter, defaultdict


LAST_NAME = "Barrientos"
SEED_NUM = 4
FAVORITE_ARTIST = "Mac DeMarco"


sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]

categories = [
    "TEMPERATURE",
    "VOLTAGE",
    "CURRENT",
    "PRESSURE",
    "SPEED"
]


artist_score = sum(
    (i + 1) * ord(char)
    for i, char in enumerate(FAVORITE_ARTIST)
)

records = []

for i, source in enumerate(sources):
    for j, category in enumerate(categories):
        value = SEED_NUM + artist_score + (i + 1) * 10 + (j + 1) * 5

        records.append({
            "Source": source,
            "Category": category,
            "Value": value
        })


records.append(records[0].copy())


records.append({
    "Source": sources[0],
    "Category": "TEMPERATURE",
    "Value": records[0]["Value"] + 5
})

records.append({
    "Source": sources[1],
    "Category": "VOLTAGE",
    "Value": records[6]["Value"] + 10
})


print("1. GENERATED TEST DATA")
for record in records:
    print(record)


organized = defaultdict(list)

for record in records:
    organized[record["Source"]].append(record)

print("\n2. ORGANIZED TEST DATA")
for source, data in organized.items():
    print(source, data)


unique_records = []

for record in records:
    if record not in unique_records:
        unique_records.append(record)

pair_counts = Counter(
    (record["Source"], record["Category"])
    for record in records
)

print("\n3. REPEATED/DISTINCT DATA")
print("Total records:", len(records))
print("Distinct records:", len(unique_records))
print("Exact duplicate records:", len(records) - len(unique_records))

print("Repeated source/category pairs:")
for pair, count in pair_counts.items():
    if count > 1:
        print(pair, "appears", count, "times")


print("\n4. NUMERICAL AND CATEGORICAL SUMMARIES")

for category in categories:
    values = [
        record["Value"]
        for record in records
        if record["Category"] == category
    ]

    print(
        category,
        "| Count:", len(values),
        "| Total:", sum(values),
        "| Average:", round(sum(values) / len(values), 2),
        "| Minimum:", min(values),
        "| Maximum:", max(values)
    )

print("\nRecords per source:")
for source, count in Counter(
    record["Source"] for record in records
).items():
    print(source, count)


print("\n5. PROCESSED RESULTS")

for category in categories:
    values = [
        record["Value"]
        for record in records
        if record["Category"] == category
    ]

    print(
        category,
        "average =",
        round(sum(values) / len(values), 2)
    )


print("\n6. ORDERED RESULTS")

ordered = sorted(
    records,
    key=lambda record: (
        record["Source"],
        record["Category"],
        record["Value"]
    )
)

for record in ordered:
    print(record)


print("\n7. FINAL SUMMARY")
print("Student:", LAST_NAME)
print("Seed number:", SEED_NUM)
print("Favorite artist:", FAVORITE_ARTIST)
print("Number of sources:", len(sources))
print("Number of categories:", len(categories))
print("Total measurements:", len(records))
print("Distinct records:", len(unique_records))
print("Exact duplicate records:", len(records) - len(unique_records))
print("Categories:", ", ".join(categories))
print("Final dataset organized and summarized successfully.")

print("\nFINAL ORGANIZED AND SUMMARIZED TEST DATA SET")

final_dataset = defaultdict(list)

for record in unique_records:
    final_dataset[record["Source"]].append(record)

for source, data in sorted(final_dataset.items()):
    print(f"\nSource: {source}")

    for record in sorted(
        data,
        key=lambda r: (r["Category"], r["Value"])
    ):
        print(
            f"  Category: {record['Category']}, "
            f"Value: {record['Value']}"
        )

print("\nDataset successfully organized and summarized.")
```
